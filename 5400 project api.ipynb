```python
#microsoft alphabet meta
from newsapi import NewsApiClient
import pandas as pd

newsapi = NewsApiClient(api_key="e50e3a3dd988450c830e80887853a65c")

articles = newsapi.get_everything(
    q="Microsoft OR Meta OR Alphabet",
    language='en',
    sort_by="publishedAt",
    from_param="2025-11-01",
    page_size=100
)

data = []
for a in articles["articles"]:
    data.append({
        "title": a["title"],
        "description": a["description"],
        "content": a["content"],
        "publishedAt": a["publishedAt"],
        "source": a["source"]["name"]
    })

df = pd.DataFrame(data)
print(df.head())
```

```python
def detect_company(text):
    t = str(text).lower()
    if "microsoft" in t:
        return "Microsoft"
    elif "meta" in t:
        return "Meta"
    elif "alphabet" in t or "google" in t:
        return "Alphabet"
    else:
        return "Unknown"

df["company"] = df["content"].apply(detect_company)
```
